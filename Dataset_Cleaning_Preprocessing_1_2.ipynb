```python
import pandas as pd
import numpy as np

print("Dataset Cleaning & Preprocessing Task 1.2 Started Successfully!")
```

```python
df = pd.read_csv('/content/student_performance_raw.csv')

df.head()
```

```python
df.isnull().sum()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Marks'] = df['Marks'].fillna(df['Marks'].median())

print("Missing values handled successfully!")
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Duplicate rows removed successfully!")
print("Remaining rows:", len(df))
```

```python
# Remove extra spaces
df['Name'] = df['Name'].str.strip()
df['Gender'] = df['Gender'].str.strip().str.title()
df['Department'] = df['Department'].str.strip().str.upper()
df['City'] = df['City'].str.strip().str.title()

print("Inconsistent formatting cleaned successfully!")
print("\nCleaned formatting preview:")
print(df[['Name', 'Gender', 'Department', 'City']].head())
```

```python
df['Attendance'] = df['Attendance'].str.replace('%', '', regex=False).astype(float)

print("Attendance column cleaned successfully!")
print(df[['Student_ID', 'Attendance']].head())
```

```python
print("Final Dataset Information:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
df.to_csv('student_performance_cleaned.csv', index=False)

print("Cleaned dataset saved successfully!")
print("File name: student_performance_cleaned.csv")
```
